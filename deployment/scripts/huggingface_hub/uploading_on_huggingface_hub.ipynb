```python
from huggingface_hub import login, HfApi
import os
```

```python
LOCAL_MODELS_DIR = "C:/Users/98922/Documents/python_scripts/AI/apps/Lung Disease Detection/deployment/saved_models"  # <-- adjust this!
DISEASE_DIR = os.path.join(LOCAL_MODELS_DIR, "diseases")
SEGMENT_DIR = os.path.join(LOCAL_MODELS_DIR, "segmentation")
HEALTHY_UNHEALTHY_DIR = os.path.join(LOCAL_MODELS_DIR, "healthy_unhealthy")
TOKEN_DIR = "./hugging_face_token.txt"

HF_USERNAME = "a1mohamadd"               # your Hugging Face username
REPO_NAME = "lung-disease-detection"
REPO_ID = f"{HF_USERNAME}/{REPO_NAME}"
```

```python
# ------------------------------------------------------------
# 1. Log in with your Hugging Face token (write permission)
# ------------------------------------------------------------
# Either set the token as an environment variable:
# Read token from file
with open("./hugging_face_token.txt", "r") as f:
    token = f.read().strip() 

if token:
    login(token=token)
else:
    # This will prompt you to paste the token interactively
    login()
```

```python
# ------------------------------------------------------------
# 3. Upload the whole folder tree
# ------------------------------------------------------------
api = HfApi()

# upload_folder:
#   - folder_path: local directory to upload
#   - repo_id: target repository
#   - repo_type: "model" (default)
#   - path_in_repo: "" means upload at root (preserves subfolders)
#   - commit_message: optional

api.upload_folder(
    folder_path=DISEASE_DIR,
    repo_id=REPO_ID,
    repo_type="model",
    path_in_repo="",          # keep top‑level structure (diseases/, healthy/, etc.)
    commit_message="Upload all models with metadata and folder structure"
)

print("\n✅ All files uploaded successfully!")
```

```python
api.upload_folder(
    folder_path=SEGMENT_DIR,
    repo_id=REPO_ID,
    repo_type="model",
    path_in_repo="",          # keep top‑level structure (diseases/, healthy/, etc.)
    commit_message="Upload segmentation files"
)

print("\n✅ Segmentation model uploaded successfully!")
```

```python
with open(TOKEN_DIR, "r") as f:
    token = f.read().strip()

login(token=token)
```

```python
api = HfApi()
```

```python
for folder_name in FOLDERS:
    local_folder = os.path.join(LOCAL_MODELS_DIR, folder_name)
    api.upload_folder(
        folder_path=local_folder,
        repo_id=REPO_ID,
        repo_type="model",
        path_in_repo=folder_name,
        allow_patterns="**/*.onnx",
        commit_message="Adding ONNX models for deployment"
    )
    print(f"✅ Uploaded all ONNX files inside {folder_name}/")
```
